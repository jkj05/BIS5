```python
import random

def binary_to_int(chromosome):
    return int("".join(map(str, chromosome)), 2)

def compute_fitness(chromosome):
    x = binary_to_int(chromosome)
    return x ** 2

def initialize_population(pop_size, chrom_len):
    population = []
    for i in range(pop_size):
        chromosome = [random.randint(0, 1) for t in range(chrom_len)]
        population.append(chromosome)
    return population

def roulette_wheel_selection(population, fitness_scores):
    total_fitness = sum(fitness_scores)
    if total_fitness == 0:
        return random.choices(population, k=len(population))
    return random.choices(population, weights=fitness_scores, k=len(population))

def crossover(parent1, parent2, crossover_rate):
    if random.random() < crossover_rate:
        point = random.randint(1, len(parent1) - 1)
        child1 = parent1[:point] + parent2[point:]
        child2 = parent2[:point] + parent1[point:]
        return child1, child2
    return parent1.copy(), parent2.copy()

def mutate(chromosome, mutation_rate):
    for i in range(len(chromosome)):
        if random.random() < mutation_rate:
            chromosome[i] = 1 - chromosome[i]
    return chromosome

def run_genetic_algorithm():
    POP_SIZE = 6
    CHROM_LEN = 5
    GENERATIONS = 20
    CROSSOVER_RATE = 0.8
    MUTATION_RATE = 0.05

    population = initialize_population(POP_SIZE, CHROM_LEN)

    print("=== INITIAL RANDOM POPULATION ===")
    for idx, chrom in enumerate(population, 1):
        x_val = binary_to_int(chrom)
        fit = compute_fitness(chrom)
        print(f"String {idx}: {chrom} | Decoded X = {x_val:2d} | Fitness = {fit:3d}")
    print("=================================\n")

    global_best_chrom = None
    global_best_fitness = -1

    print("=== GENETIC ALGORITHM FOR F(X) = X^2 ===")

    for gen in range(1, GENERATIONS + 1):
        fitness_scores = [compute_fitness(chrom) for chrom in population]

        current_best_idx = fitness_scores.index(max(fitness_scores))
        current_best_chrom = population[current_best_idx]
        current_best_fitness = fitness_scores[current_best_idx]

        if current_best_fitness > global_best_fitness:
            global_best_fitness = current_best_fitness
            global_best_chrom = current_best_chrom.copy()

        print(f"Generation {gen:02d}: Best X = {binary_to_int(current_best_chrom):2d} | Best Fitness = {current_best_fitness:3d}")

        if global_best_fitness == 961:
            print(f"\n[!] Global Optimum reached at Generation {gen}!")
            break

        mating_pool = roulette_wheel_selection(population, fitness_scores)
        next_population = []

        for i in range(0, POP_SIZE, 2):
            p1 = mating_pool[i]
            p2 = mating_pool[i+1]

            c1, c2 = crossover(p1, p2, CROSSOVER_RATE)

            next_population.append(mutate(c1, MUTATION_RATE))
            next_population.append(mutate(c2, MUTATION_RATE))

        population = next_population

    print("\n=== FINAL RESULT ===")
    print(f"Optimal Binary Chromosome : {global_best_chrom}")
    print(f"Decoded X Integer Value    : {binary_to_int(global_best_chrom)}")
    print(f"Maximum Fitness Achieved   : {global_best_fitness}")

if __name__ == "__main__":
    run_genetic_algorithm()
```
